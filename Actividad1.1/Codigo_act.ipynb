Paso 3- Convertir los datos de la variable “Potencial de compra” a tipo float

```python
df_filtrado['Potencial de compra'] = df_filtrado['Potencial de compra'].astype(str)
df_filtrado['Potencial de compra'] = df_filtrado['Potencial de compra'].str.replace("Ventas ", "0", regex=False)
df_filtrado['Potencial de compra'] = df_filtrado['Potencial de compra'].str.replace("%", "", regex=False)
df_filtrado['Potencial de compra'] = df_filtrado['Potencial de compra'].str.replace("'", "", regex=False)
df_filtrado['Potencial de compra'] = df_filtrado['Potencial de compra'].str.replace("&", "", regex=False)
df_filtrado['Potencial de compra'] = df_filtrado['Potencial de compra'].replace("nan", np.nan)

df_filtrado['Potencial de compra'] = df_filtrado['Potencial de compra'].astype(float)
df_filtrado.info()
```

Paso 4- Obtener el CSV General y nombrarlo “Citas_Digital_General.csv”

```python
df_filtrado.to_csv("Citas_Digital_General.csv", index=False)
```

Paso 5- A partir del archivo “Citas_Digital_General.csv” realizar un filtro por filas para obtener un archivo .csv por cada mes (Desde abril 2025 hasta julio 2026) y nombrarlos con el mes y año correspondiente ejemplo: “Abril_2025.csv” (16 entregables)

```python
df_general = pd.read_csv("Citas_Digital_General.csv")

#Rangos de filas segun las filas marcadoras de cada mes en el Excel original
#(fila_inicio, fila_fin) son indices de df_general (ya excluyen filas marcadoras y filas vacias)
rangos_meses = [
    ("Abril",      "2025", 0,   30),
    ("Mayo",       "2025", 34,  58),
    ("Junio",      "2025", 61,  86),
    ("Julio",      "2025", 88,  120),
    ("Agosto",     "2025", 122, 166),
    ("Septiembre", "2025", 168, 230),
    ("Octubre",    "2025", 232, 290),
    ("Noviembre",  "2025", 292, 343),
    ("Diciembre",  "2025", 345, 391),
    ("Enero",      "2026", 393, 428),
    ("Febrero",    "2026", 430, 460),
    ("Marzo",      "2026", 462, 477),
    ("Abril",      "2026", 479, 533),
    ("Mayo",       "2026", 540, 598),
    ("Junio",      "2026", 600, 640),
    ("Julio",      "2026", 642, 685),
]
```

```python
for mes, anio, fila_ini, fila_fin in rangos_meses:
    #Filtro por filas: seleccionamos el bloque de filas correspondiente al mes
    filtro_mes = df_general.iloc[fila_ini:fila_fin + 1]

    filtro_mes.to_csv(f"{mes}_{anio}.csv", index=False)
    print(f"{mes}_{anio}.csv -> {len(filtro_mes)} filas")
```
